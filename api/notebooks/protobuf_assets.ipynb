Comparaison JSON & protobuf

```python
sample_data = {
    "name": "Monja",
    "age": 21,
    "email": "monja@example.com",
    "phones": ["0340861263", "0330861263"]
}
iterations = 1_000_000
```

mesurer JSON

```python
start_json = time.time()
for _ in tqdm(range(iterations)):
    s = json.dumps(sample_data)
    obj = json.loads(s)
json_time = time.time() - start_json
```

mesurer proto

```python
start_proto = time.time()
for _ in tqdm(range(iterations)):
    person = person_pb2.Person(
        name=sample_data["name"],
        email=sample_data["email"],
        phones=sample_data["phones"]
    )
    data = person.SerializeToString()
    p2 = person_pb2.Person()
    p2.ParseFromString(data)
proto_time = time.time() - start_proto
```

resultat

```python
print(f"JSON (serialization+deserialization) : {json_time:.6f}s")
print(f"Protobuf (serialization+deserialization) : {proto_time:.6f}s")
print(f"difference : {(json_time - proto_time):.6f}s")
print(f"protobuf = x{json_time/proto_time:.6f} json")
```
